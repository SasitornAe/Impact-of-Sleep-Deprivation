```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. โหลดข้อมูล
df = pd.read_csv(r"C:\Users\Sasitorn Pansorn\Downloads\Project ไพธอน\Sleep_Deprivation.csv")

# 2. ตั้งค่า Figure และสีพื้นหลัง
fig, axes = plt.subplots(1, 2, figsize=(14, 7), facecolor='#f8f9fa')

# 3. กำหนด Palette สีแบบกำหนดเอง
colors_gender = ['#6baed6', '#fc9272']
colors_bmi = ['#a1d99b', '#9ecae1', '#fdae6b']

# --- Subplot 1: 3D-like Pie Chart (เพศ) ---
gender_counts = df['Gender'].value_counts()
wedges1, texts1, autotexts1 = axes[0].pie(
    gender_counts, 
    labels=gender_counts.index, 
    autopct='%1.1f%%', 
    startangle=140,
    colors=colors_gender,
    explode=(0.05, 0),       # ดึงชิ้นส่วนแรกออกมาเล็กน้อย
    shadow=True,             # เพิ่มเงา 3D
    textprops=dict(color="black", size=11, weight="bold")
)
axes[0].set_title('Participant Gender Ratio', fontsize=14, fontweight='bold', pad=15)

# --- Subplot 2: Donut Chart (กลุ่ม BMI) ---
bmi_counts = df['bmi_category'].value_counts()
wedges2, texts2, autotexts2 = axes[1].pie(
    bmi_counts, 
    labels=bmi_counts.index, 
    autopct='%1.1f%%', 
    startangle=90,
    colors=colors_bmi,
    explode=(0.03, 0.03, 0.03), # แยกทุกชิ้นส่วนออกนิดหน่อย
    pctdistance=0.8,
    wedgeprops=dict(width=0.4, edgecolor='white', linewidth=2) # width=0.4 ทำให้กลายเป็นโดนัท
)
axes[1].set_title('BMI Category Breakdown (Donut)', fontsize=14, fontweight='bold', pad=15)

# 4. ปรับแต่งตัวอักษรเปอร์เซ็นต์ข้างในกราฟ
for autotext in autotexts1:
    autotext.set_color('white')
    autotext.set_fontsize(12)

for autotext in autotexts2:
    autotext.set_fontsize(10)
    autotext.set_weight('bold')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. อ่านไฟล์ข้อมูล
df = pd.read_csv(r"C:\Users\Sasitorn Pansorn\Downloads\Project ไพธอน\Sleep_Deprivation.csv")

# 2. คำนวณค่าเฉลี่ย PVT Reaction Time แยกตามกลุ่ม BMI
bmi_reaction = df.groupby('bmi_category')['PVT_Reaction_Time'].mean().reset_index()

plt.figure(figsize=(8, 5), facecolor='#ffffff')

# 3. วาด Bar Chart
ax = sns.barplot(
    data=bmi_reaction,
    x='bmi_category',
    y='PVT_Reaction_Time',
    hue='bmi_category',
    palette='Blues_r'
)

# 4. ตกแต่งหัวข้อและแกน
plt.title('Average PVT Reaction Time by BMI Category', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('BMI Category', fontsize=11, fontweight='bold')
plt.ylabel('Mean PVT Reaction Time (ms)', fontsize=11, fontweight='bold')

# ซ่อน Legend และขอบกราฟด้านบน/ขวา
if ax.get_legend():
    ax.get_legend().remove()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 5. ใส่ตัวเลขกำกับค่าเฉลี่ยไว้บนยอดแท่งกราฟ
for p in ax.patches:
    ax.annotate(
        f'{p.get_height():.1f} ms',
        (p.get_x() + p.get_width() / 2., p.get_height()),
        ha='center', va='center',
        xytext=(0, 8),
        textcoords='offset points',
        fontsize=10, fontweight='bold', color='#2c3e50'
    )

plt.tight_layout()
plt.show()
```

```python
import time
import numpy as np
import pandas as pd

# 1. อ่านข้อมูลจากไฟล์ CSV
df = pd.read_csv('Sleep_Deprivation.csv')

# ดึงข้อมูลคอลัมน์ตัวเลขมาแปลงเป็น NumPy Array
sleep_hours = df['Sleep_Hours'].values
quality_score = df['Sleep_Quality_Score'].values

# --- 1) วิธีที่ 1: แบบใช้ For Loop (ช้ากว่า และโค้ดยาวกว่า) ---
start_time = time.time()
result_loop = [
    sleep_hours[i] * quality_score[i] for i in range(len(sleep_hours))
]
time_loop = (time.time() - start_time) * 1000  # แปลงเป็นมิลลิวินาที

# --- 2) วิธีที่ 2: แบบใช้ NumPy Vectorized (เร็วกว่า และเขียนสั้นมาก) ---
start_time = time.time()
result_vector = sleep_hours * quality_score
time_vector = (time.time() - start_time) * 1000  # แปลงเป็นมิลลิวินาที

# แสดงผลลัพธ์การวัดเวลา
print(f'เวลาที่ใช้แบบ For Loop: {time_loop:.4f} มิลลิวินาที')
print(f'เวลาที่ใช้แบบ Vectorized: {time_vector:.4f} มิลลิวินาที')
print('ผลลัพธ์การคำนวณ (Vectorized):', result_vector[:5])
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. อ่านข้อมูลจากไฟล์ CSV
df = pd.read_csv('Sleep_Deprivation.csv')

# 2. ตั้งค่าขนาดและสไตล์ของกราฟ
plt.figure(figsize=(10, 6), dpi=300)

# 3. พลอต Scatter Plot โดยแบ่งสีตามเพศ (Gender)
genders = df['Gender'].unique()
colors = {'Male': '#3498db', 'Female': '#e74c3c'}  # ฟ้าสำหรับชาย, แดงสำหรับหญิง

for gender in genders:
  subset = df[df['Gender'] == gender]
  plt.scatter(
      subset['Sleep_Hours'],
      subset['N_Back_Accuracy'],
      color=colors.get(gender, '#2ecc71'),
      label=gender,
      alpha=0.75,
      s=70,
      edgecolors='white',
      linewidth=0.8,
  )

# 4. ตกแต่งองค์ประกอบกราฟ
plt.title(
    'Relationship Between Sleep Hours and N-Back Accuracy',
    fontsize=14,
    fontweight='bold',
    pad=15,
)
plt.xlabel('Sleep Hours (Hours)', fontsize=12, labelpad=10)
plt.ylabel('N-Back Accuracy Score (%)', fontsize=12, labelpad=10)

# เพิ่มเส้นแนวโน้ม (Trendline) ใช้ np.poly1d ที่ถูกต้อง
z = np.polyfit(df['Sleep_Hours'], df['N_Back_Accuracy'], 1)
p = np.poly1d(z)
# จัดเรียงค่า X เพื่อให้เส้นแนวโน้มลากเชื่อมต่อกันอย่างสวยงาม
x_sorted = np.sort(df['Sleep_Hours'])
plt.plot(
    x_sorted,
    p(x_sorted),
    color='gray',
    linestyle='--',
    linewidth=2,
    alpha=0.7,
    label='Trend',
)

# ตั้งค่ากล่องคำอธิบายและเส้นตาราง
plt.legend(frameon=True, facecolor='white', edgecolor='none', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

# 5. บันทึกภาพความละเอียดสูง
plt.savefig('beautiful_scatter_sleep.png', dpi=300)
print("บันทึกภาพกราฟสำเร็จในชื่อ 'beautiful_scatter_sleep.png'")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. อ่านข้อมูลจากไฟล์ CSV
df = pd.read_csv('Sleep_Deprivation.csv')

# ดึงข้อมูลคอลัมน์ Sleep_Hours (x) และ PVT_Reaction_Time (y) มาคำนวณ
x = df['Sleep_Hours'].values
y = df['PVT_Reaction_Time'].values

# --- คำนวณทางสถิติด้วย NumPy ---
# 1. เส้นตรงกำลังสองน้อยที่สุด (Linear Regression / Polyfit ดีกรี 1)
m, b = np.polyfit(x, y, 1)
print(f'สมการเส้นตรง: y = {m:.3f}x + {b:.3f}')

# 2. สัมประสิทธิ์สหสัมพันธ์ (Correlation Coefficient)
r = np.corrcoef(x, y)[0, 1]
print(f'ค่า Correlation (r): {r:.4f}')

# 3. ควอร์ไทล์ (Percentiles: 25%, 50%, 75%)
q25, q50, q75 = np.percentile(y, [25, 50, 75])
print(f'Quartiles (Q1, Median, Q3): {q25:.1f}, {q50:.1f}, {q75:.1f}')

# --- สร้างกราฟแสดงผล 2 ช่อง (ซ้าย: Scatter + Polyfit, ขวา: Histogram + Percentiles) ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), dpi=300)

# กราฟซ้าย: np.polyfit & r
ax1.scatter(x, y, color='mediumpurple', s=30, alpha=0.8)
x_sorted = np.sort(x)
ax1.plot(
    x_sorted,
    m * x_sorted + b,
    color='darkblue',
    linewidth=2,
    label=f'y={m:.2f}x+{b:.2f}',
)
ax1.set_title(
    f'np.polyfit | r = {r:.3f}', fontsize=12, fontweight='bold', color='navy'
)
ax1.legend(loc='upper right')
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.set_xlabel('Sleep Hours')
ax1.set_ylabel('PVT Reaction Time')

# กราฟขวา: np.percentile (ฮิสโตแกรมพร้อมเส้นแสดง Q1, Median, Q3)
ax2.hist(
    y, bins=20, color='cornflowerblue', alpha=0.4, edgecolor='white'
)
ax2.axvline(
    q25, color='c', linestyle='--', linewidth=2, label=f'Q1={q25:.1f}'
)
ax2.axvline(
    q50,
    color='deeppink',
    linestyle='--',
    linewidth=2,
    label=f'Median={q50:.1f}',
)
ax2.axvline(
    q75, color='orange', linestyle='--', linewidth=2, label=f'Q3={q75:.1f}'
)
ax2.set_title(
    'np.percentile', fontsize=12, fontweight='bold', color='navy'
)
ax2.legend(loc='upper right')
ax2.grid(True, linestyle='--', alpha=0.5)
ax2.set_xlabel('PVT Reaction Time')
ax2.set_ylabel('Frequency')

plt.tight_layout()

# บันทึกภาพกราฟเก็บไว้
plt.savefig('numpy_statistics_chart.png', dpi=300)
print("บันทึกภาพสำเร็จในชื่อ 'numpy_statistics_chart.png'")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. อ่านข้อมูลจากไฟล์ CSV
df = pd.read_csv('Sleep_Deprivation.csv')

# 2. จัดกลุ่มข้อมูล (Group By) ตามชั่วโมงการนอน (ปัดเศษให้เป็นเลขกลมๆ เพื่อให้เส้นกราฟเรียบเนียน)
df['Sleep_Rounded'] = df['Sleep_Hours'].round(0)
grouped = (
    df.groupby('Sleep_Rounded')[
        ['N_Back_Accuracy', 'Emotion_Regulation_Score']
    ]
    .mean()
    .reset_index()
)

# กรองเฉพาะช่วงที่มีข้อมูลครบถ้วน
grouped = grouped.sort_values('Sleep_Rounded')

x = grouped['Sleep_Rounded'].values
y1 = grouped['N_Back_Accuracy'].values
y2 = grouped['Emotion_Regulation_Score'].values * 1.2  # ปรับสเกลให้อยู่ระดับเดียวกัน

# 3. สร้างกราฟ
plt.figure(figsize=(10, 5), dpi=300)

# พลอตเส้นกราฟแบบเรียบเนียนขึ้น (Smooth line หรือเส้นตรงปกติที่ผ่านการเฉลี่ยแล้ว)
plt.plot(
    x, y1, marker='o', label='Avg N-Back Accuracy', color='#3949ab', linewidth=2
)
plt.plot(
    x,
    y2,
    '--s',
    label='Avg Emotion Score (Scaled)',
    color='#ec407a',
    linewidth=2,
)

# ระบายพื้นที่ระหว่างเส้น
plt.fill_between(x, y1, y2, color='#7e57c2', alpha=0.2)

# 4. ตกแต่งองค์ประกอบกราฟ
plt.title(
    'Comparison of Average Scores by Rounded Sleep Hours',
    fontsize=13,
    fontweight='bold',
    color='navy',
)
plt.xlabel('Sleep Hours (Rounded)', fontsize=11)
plt.ylabel('Average Score', fontsize=11)
plt.legend(loc='upper left', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

# 5. บันทึกภาพ
plt.savefig('smooth_filled_chart.png', dpi=300)
print('บันทึกภาพกราฟแบบจัดกลุ่มสวยงามสำเร็จ!')

plt.show()
```

```python
import pandas as pd
import plotly.express as px

# 1. อ่านข้อมูลจากไฟล์ CSV
df = pd.read_csv('Sleep_Deprivation.csv')

# --- 1. กราฟวงกลมแบบโดนัท (Donut Chart) ---
# นับจำนวนสัดส่วนผู้เข้าร่วมตามประเภท BMI (bmi_category)
bmi_counts = df['bmi_category'].value_counts().reset_index()
bmi_counts.columns = ['bmi_category', 'count']

# สร้างกราฟวงกลมแบบมีรูตรงกลาง (hole=0.42) ตามตัวอย่าง
fig_pie = px.pie(
    bmi_counts,
    names='bmi_category',
    values='count',
    hole=0.42,
    title='สัดส่วนกลุ่มดัชนีมวลกาย (bmi_category)',
)
fig_pie.show()


# --- 2. แผนภาพความร้อน (Heatmap Correlation) ---
# เลือกคอลัมน์ตัวเลขที่น่าสนใจมาคำนวณหาความสัมพันธ์ (corr)
cols = [
    'Sleep_Hours',
    'Sleep_Quality_Score',
    'N_Back_Accuracy',
    'PVT_Reaction_Time',
    'Age',
]
corr = df[cols].corr()

# สร้างแผนภาพความร้อนด้วย px.imshow ตามตัวอย่างเป๊ะๆ
fig_heatmap = px.imshow(
    corr,
    text_auto='.2f',  # แสดงตัวเลขทศนิยม 2 ตำแหน่งในช่อง
    color_continuous_scale='RdBu',  # ใช้โทนสีแดง-น้ำเงิน
    zmin=-1,
    zmax=1,
    title='แผนภาพความร้อนแสดงความสัมพันธ์ตัวแปร (Heatmap Correlation)',
)
fig_heatmap.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. อ่านข้อมูล
df = pd.read_csv(r"C:\Users\Sasitorn Pansorn\Downloads\Project ไพธอน\Sleep_Deprivation.csv")

# 2. จัดกลุ่มข้อมูลหาค่าเฉลี่ยชั่วโมงการนอนตามประเภท BMI
avg_df = df.groupby('bmi_category')['Sleep_Hours'].mean().reset_index()

# เรียงลำดับหมวดหมู่ BMI ให้สวยงาม (ถ้าต้องการ) เช่น Normal, Overweight, Obesity
category_order = ['Normal', 'Overweight', 'Obesity']
# กรองและจัดเรียงตามลำดับ (หากมีครบทุกกลุ่ม)
avg_df['bmi_category'] = pd.Categorical(avg_df['bmi_category'], categories=category_order, ordered=True)
avg_df = avg_df.sort_values('bmi_category').reset_index(drop=True)

# 3. กำหนดขนาดและสไตล์ของกราฟ
plt.figure(figsize=(10, 6), facecolor='#ffffff')
sns.set_theme(style="whitegrid")

# ใช้สีโทนฟ้าสวยงามจาก seaborn palette
colors = sns.color_palette("Blues_d", len(avg_df))

# พลอตราฟแท่ง
bars = plt.bar(
    avg_df['bmi_category'],
    avg_df['Sleep_Hours'],
    color=colors,
    width=0.45,
    edgecolor='#1d3557',
    linewidth=1,
    alpha=0.85
)

# 4. ตกแต่งชื่อและป้ายกำกับ
plt.title('Average Sleep Hours by BMI Category', fontsize=14, fontweight='bold', pad=20, color='#1d3557')
plt.xlabel('BMI Category', fontsize=11, fontweight='bold', labelpad=10, color='#333333')
plt.ylabel('Average Sleep Hours (Hours)', fontsize=11, fontweight='bold', labelpad=10, color='#333333')

# เปิด Grid ด้านหลังเฉพาะแนวตั้ง (Y-axis)
plt.grid(axis='y', linestyle=':', alpha=0.7)

# ปรับให้แกน Y เริ่มจาก 0 และเผื่อพื้นที่ด้านบนสำหรับตัวเลข
max_val = avg_df['Sleep_Hours'].max()
plt.ylim(0, max_val + 1.2)

# แสดงตัวเลขกำกับไว้บนแท่งกราฟแต่ละแท่งให้อ่านง่าย
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.15,
        f'{height:.2f} hrs',
        ha='center',
        va='bottom',
        fontsize=10,
        fontweight='bold',
        color='#1d3557'
    )

# ซ่อนขอบบนและขวาเพื่อความสะอาดเรียบร้อย
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()

# 5. บันทึกภาพความละเอียดสูงและแสดงผล
plt.savefig('beautiful_sleep_bmi_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import time
import matplotlib.pyplot as plt

sizes = [10_000, 100_000, 1_000_000]
loop_ms, vec_ms = [], []

for n in sizes:
    a = np.random.rand(n)
    b = np.random.rand(n)

    # แบบวนลูป
    t = time.perf_counter()
    c = [a[i] * b[i] for i in range(len(a))]
    loop_ms.append((time.perf_counter() - t) * 1000)

    # แบบ vectorized (วัดหลายรอบแล้วเฉลี่ย เพราะเร็วมาก)
    reps = 20
    t = time.perf_counter()
    for _ in range(reps):
        c = a * b
    vec_ms.append((time.perf_counter() - t) * 1000 / reps)

x = np.arange(len(sizes))
w = 0.38
PINK, BLUE = '#EC4899', '#3B4CCA'

fig, ax = plt.subplots(figsize=(8, 5))
b1 = ax.bar(x - w/2, loop_ms, width=w, color=PINK, label='for loop')
b2 = ax.bar(x + w/2, vec_ms, width=w, color=BLUE, label='NumPy vectorized')
ax.bar_label(b1, fmt='%.3g', padding=2, fontsize=9)
ax.bar_label(b2, fmt='%.3g', padding=2, fontsize=9)

ax.set_yscale('log')                      # สเกล log ให้เห็นทั้งแท่งสูงและแท่งเตี้ย
ax.set_xticks(x)
ax.set_xticklabels([f'{n:,}' for n in sizes])
ax.set_xlabel('Array size')
ax.set_ylabel('Time (ms, log scale)')
ax.set_title('for loop vs NumPy', fontweight='bold', color='#2D2A8C')
ax.grid(axis='y', linestyle=':', alpha=0.6)
ax.set_axisbelow(True)
ax.spines[['top', 'right']].set_visible(False)
ax.legend()
fig.tight_layout()
plt.show()

# พิมพ์ผลเป็นตัวเลขด้วย
for n, l, v in zip(sizes, loop_ms, vec_ms):
    print(f'{n:>9,}  loop={l:8.2f} ms  numpy={v:6.2f} ms  speedup={l/v:5.1f}x')
```

```python
import warnings; warnings.filterwarnings('ignore')
import numpy as np, pandas as pd
from sklearn.model_selection import RepeatedKFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR

df = pd.read_csv('Sleep_Deprivation.csv', encoding='utf-8-sig')
num = ['Sleep_Hours','Sleep_Quality_Score','Daytime_Sleepiness','Stroop_Task_Reaction_Time',
       'N_Back_Accuracy','Emotion_Regulation_Score','PVT_Reaction_Time','Age','bmi','gender_code']
target = 'PVT_Reaction_Time'                        # เปลี่ยนตัวแปรที่จะพยากรณ์ได้
X, y = df[[c for c in num if c != target]], df[target]

models = {
 'Mean baseline': DummyRegressor(),
 'Linear':        make_pipeline(StandardScaler(), LinearRegression()),
 'Ridge':         make_pipeline(StandardScaler(), RidgeCV(alphas=np.logspace(-2, 3, 30))),
 'Lasso':         make_pipeline(StandardScaler(), LassoCV(cv=5, max_iter=20000)),
 'Decision Tree': DecisionTreeRegressor(max_depth=3, random_state=0),
 'Random Forest': RandomForestRegressor(n_estimators=60, min_samples_leaf=5, n_jobs=-1, random_state=0),
 'Gradient Boost':GradientBoostingRegressor(n_estimators=60, max_depth=2, learning_rate=0.05, random_state=0),
 'KNN':           make_pipeline(StandardScaler(), KNeighborsRegressor(10)),
 'SVR':           make_pipeline(StandardScaler(), SVR()),
}

cv = RepeatedKFold(n_splits=5, n_repeats=4, random_state=42)
for name, m in models.items():
    cv_r2 = cross_val_score(m, X, y, cv=cv, scoring='r2').mean()   # R² บนข้อมูลที่ไม่เคยเห็น
    train_r2 = m.fit(X, y).score(X, y)                              # R² บนข้อมูลฝึก
    print(f'{name:15s} CV R² = {cv_r2:6.3f}   Train R² = {train_r2:6.3f}')
```

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

df = pd.read_csv('Sleep_Deprivation.csv', encoding='utf-8-sig')
feats = ['Sleep_Hours','Sleep_Quality_Score','Daytime_Sleepiness','Stroop_Task_Reaction_Time',
         'N_Back_Accuracy','Emotion_Regulation_Score','PVT_Reaction_Time','Age','bmi']
Z = StandardScaler().fit_transform(df[feats])      # K-means ต้องปรับสเกลก่อนเสมอ
tss = ((Z - Z.mean(0))**2).sum()

ks, var_exp, sil = range(2, 9), [], []
for k in ks:
    km = KMeans(k, n_init=20, random_state=0).fit(Z)
    var_exp.append(1 - km.inertia_ / tss)           # "R²" ของการจัดกลุ่ม (0 ถึง 1)
    sil.append(silhouette_score(Z, km.labels_))
    print(f'k={k}  var_explained={var_exp[-1]:.3f}  silhouette={sil[-1]:.3f}')

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(ks, var_exp, marker='o'); axes[0].set_title('Variance explained (elbow)')
axes[1].plot(ks, sil, marker='o');     axes[1].set_title('Silhouette')
axes[1].axhline(0.25, color='gray'); axes[1].set_ylim(0, 0.6)   # เส้นเกณฑ์ 0.25
for ax in axes: ax.set_xlabel('k')
plt.tight_layout(); plt.show()
```

```python
import numpy as np, pandas as pd
from scipy import stats

df = pd.read_csv('Sleep_Deprivation.csv', encoding='utf-8-sig')
pvt, sh = df['PVT_Reaction_Time'], df['Sleep_Hours']

# ข้อ 1: เปรียบเทียบกลุ่ม BMI
g = [df.loc[df.bmi_category == c, 'PVT_Reaction_Time'] for c in ['Normal', 'Overweight', 'Obesity']]
print('Shapiro:', [round(stats.shapiro(x).pvalue, 3) for x in g], '| Levene:', round(stats.levene(*g).pvalue, 3))
F, p = stats.f_oneway(*g)
eta2 = sum(len(x) * (x.mean() - pvt.mean())**2 for x in g) / ((pvt - pvt.mean())**2).sum()
print(f'ANOVA F={F:.2f} p={p:.3f} eta2={eta2:.3f}')
print('Kruskal p=%.3f' % stats.kruskal(*g).pvalue)

# ข้อ 2: สถิติเชิงบรรยายของชั่วโมงนอน
n = len(sh); se = sh.std() / np.sqrt(n); t = stats.t.ppf(.975, n - 1)
print(f'mean={sh.mean():.2f} SD={sh.std():.2f} median={sh.median():.2f} '
      f'IQR={sh.quantile(.25):.2f}-{sh.quantile(.75):.2f} 95%CI=({sh.mean()-t*se:.2f}, {sh.mean()+t*se:.2f})')
print('Shapiro p=%.3f' % stats.shapiro(sh).pvalue)
print(pd.cut(sh, [0, 6, 8, 99], right=False, labels=['<6', '6-<8', '>=8']).value_counts(normalize=True).round(2))

# ข้อ 3: ความสัมพันธ์
r, pr = stats.pearsonr(sh, pvt); rho, ps = stats.spearmanr(sh, pvt)
z, sz = np.arctanh(r), 1 / np.sqrt(n - 3)
print(f'Pearson r={r:.2f} p={pr:.3f} CI=({np.tanh(z-1.96*sz):.2f}, {np.tanh(z+1.96*sz):.2f}) | Spearman rho={rho:.2f} p={ps:.3f}')
```
